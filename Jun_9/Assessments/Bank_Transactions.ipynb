```python
total_tx = rdd.count()
total_val = amount_rdd.sum()
avg_val = amount_rdd.mean()
freq_dist = rdd.map(lambda x: (x, 1)).reduceByKey(lambda a, b: a + b).collect()

print("=========================================")
print("       BANK TRANSACTION DASHBOARD        ")
print("=========================================")
print(f"Total Transactions Processed : {total_tx}")
print(f"Total Financial Value        : ${total_val}")
print(f"Average Transaction Value    : ${avg_val:.2f}")
print("-----------------------------------------")
print("Transaction Distribution Breakdown:")
for tx_type, count in freq_dist:
    pct = (count / total_tx) * 100
    print(f" - {tx_type:<15} : {count:>2} ({pct:>5.1f}%)")
print("=========================================")
```
