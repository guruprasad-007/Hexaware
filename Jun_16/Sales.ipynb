```python
windowSpecRevenueDesc = Window.orderBy(F.desc("total_amount"))
windowCityRevenue = Window.partitionBy("city").orderBy(F.desc("total_amount"))
windowCategoryRevenue = Window.partitionBy("category").orderBy(F.desc("total_amount"))
windowRunningTotal = Window.orderBy("order_id").rowsBetween(Window.unboundedPreceding, Window.currentRow)
```

```python
df_win = df_calc.withColumn("row_number", F.row_number().over(windowSpecRevenueDesc))
```

```python
df_win = df_win.withColumn("dense_rank", F.dense_rank().over(windowSpecRevenueDesc))
```

```python
extracted_df = df
```

```python
transformed_df = extracted_df.withColumn("total_amount", F.col("quantity") * F.col("price")) \
                              .withColumn("tax", F.col("total_amount") * 0.05) \
                              .withColumn("final_amount", F.col("total_amount") + F.col("tax"))
```

```python
filtered_df = transformed_df.filter(F.col("status") == "Completed")
```

```python
total_etl_revenue = filtered_df.select(F.sum("final_amount")).collect()[0][0]
```

```python
city_report = filtered_df.groupBy("city").agg(F.sum("final_amount").alias("city_revenue"))
```

```python
category_report = filtered_df.groupBy("category").agg(F.sum("final_amount").alias("category_revenue"))
```

```python
city_report.write.mode("overwrite").parquet("/tmp/outputs/city_revenue_report")
category_report.write.mode("overwrite").parquet("/tmp/outputs/category_revenue_report")
```

```python
read_city_report = spark.read.parquet("/tmp/outputs/city_revenue_report")
```

```python
read_city_report.createOrReplaceTempView("v_city_revenue")
```
