```python
from pyspark.sql import functions as F
```

```python
# 1. Create account data
bank_data = [
    ("ACC101", "Kiran Rao", "Hyderabad", 45000),
    ("ACC102", "Nisha Reddy", "Bangalore", 120000),
    ("ACC103", "Arjun Verma", "Hyderabad", 85000),
    ("ACC104", "Meera Nair", "Chennai", 63000),
    ("ACC105", "Farhan Ali", "Bangalore", 95000)
]
columns = ["account_no", "account_holder", "city", "balance"]
df_bank = spark.createDataFrame(bank_data, columns)
```

```python
# 2 & 3. Calculate total balance by city and display report
df_report = df_bank.groupBy("city").agg(F.sum("balance").alias("total_balance"))
display(df_report)
```
