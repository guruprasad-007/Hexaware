```python
# 1 & 2. Create product dataset
products = [
    ("P001", "Laptop", "Electronics", 65000),
    ("P002", "Smartphone", "Electronics", 35000),
    ("P003", "Desk Chair", "Furniture", 8500),
    ("P004", "Monitor", "Electronics", 15000),
    ("P005", "Dining Table", "Furniture", 22000)
]
columns = ["product_id", "product_name", "category", "price"]
df_prod = spark.createDataFrame(products, columns)
```

```python
# 3. Display products
display(df_prod)
```

```python
# 4. Create a temp view
df_prod.createOrReplaceTempView("view_products")
```

```python
# 5 & 6. Calculate and display average product price using SQL
avg_price_df = spark.sql("""
    SELECT category, AVG(price) as avg_price 
    FROM view_products 
    GROUP BY category
""")
display(avg_price_df)
```
