```python
import os
from datetime import datetime
from airflow import DAG
from airflow.providers.standard.operators.python import PythonOperator

def create_result_file():
    file_path = "/tmp/results.txt"
    os.makedirs(os.path.dirname(file_path), exist_ok=True)
    content = "Rahul,Pass\nPriya,Fail\nAmit,Pass\nSneha,Pass\nKiran,Fail\nMegha,Pass"
    with open(file_path, "w") as f:
        f.write(content.strip())
    print(f"✔ Created {file_path}")

def count_pass_fail():
    file_path = "/tmp/results.txt"
    pass_count = 0
    fail_count = 0
    with open(file_path, "r") as f:
        for line in f:
            if line.strip():
                _, result = line.strip().split(",")
                if result.strip().lower() == "pass": pass_count += 1
                else: fail_count += 1
    print(f"Calculated: Pass={pass_count}, Fail={fail_count}")
    return pass_count, fail_count

def generate_result_summary():
    pass_count, fail_count = count_pass_fail()
    report_path = "/tmp/result_summary.txt"
    with open(report_path, "w") as f:
        f.write(f"Total Pass = {pass_count}\nTotal Fail = {fail_count}\n")
    print(f"✔ Generated Summary at {report_path}")
    print("\n📋 File Content:")
    with open(report_path, "r") as f: print(f.read())

with DAG(dag_id='exercise_9_exam_results', start_date=datetime(2026, 1, 1), schedule=None, catchup=False) as dag:
    t1 = PythonOperator(task_id='create_result_file', python_callable=create_result_file)
    t2 = PythonOperator(task_id='count_pass_fail', python_callable=count_pass_fail)
    t3 = PythonOperator(task_id='generate_result_summary', python_callable=generate_result_summary)
    t1 >> t2 >> t3

# Trigger in Colab
create_result_file()
generate_result_summary()
```
