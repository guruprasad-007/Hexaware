```python
df_categorized = df.withColumn("order_value_category", 
    when(df["total_amount"] >= 50000, "High")
    .when(df["total_amount"] >= 20000, "Medium")
    .otherwise("Low")
)
df_categorized.show()
```

```python
from pyspark.sql.functions import when

df.withColumn("order_value_category",
    when(df["total_amount"] >= 50000, "High")
    .when(df["total_amount"] >= 20000, "Medium")
    .otherwise("Low")
).groupBy("order_value_category").count().show()
```

```python
df.withColumn("discount", df["total_amount"] * 0.10) \
  .withColumn("final_amount", col("total_amount") - col("discount")) \
  .filter(df["order_status"] == "Delivered") \
  .select("order_id", "customer_name", "product", "final_amount", "order_status") \
  .show()
```
